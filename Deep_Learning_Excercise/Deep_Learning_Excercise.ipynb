# 1. Basic Concepts and Tensors Operation

### Tensor Dimension Management and Reshaping 

In Pytorch format for images are:

(N, C, H, W)

- N → batch size
- C → channels
- H → height
- W → width

Created a tensor named images representing a batch of 32 RGB images.

Tensor dimensions: (32, 3, 64, 64).

```python
images = torch.randn(32, 3, 64, 64)

print(images.shape)
```

```python
images.numel()
```

Reshaped the image tensor into a 2D tensor suitable for a fully connected layer.

```python
flatten_images = images.reshape(32, 12288)
flatten_images.shape
```

Used the permute() function to rearrange tensor dimensions for visualization.

```python
first_image = images[0]
image_for_plot = first_image.permute(1, 2, 0)
image_for_plot.shape
```
